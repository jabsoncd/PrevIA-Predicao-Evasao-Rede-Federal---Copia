```python
import requests
import json

# URL base da API
BASE_URL = "http://localhost:8000"
```

```python
# Exemplo de dados para predição individual
dados_estudante = {
    "cor_raca": "Parda",
    "idade": 20,
    "sexo": "Feminino",
    "renda_familiar": "0<RFP<=0,5",
    "modalidade_de_ensino": "Educação Presencial",
    "tipo_de_oferta": "Integrado",
    "turno": "Vespertino",
    "nome_de_curso": "Técnico em Informática",
    "eixo_tecnologico_escolhido": "Informação e Comunicação",
    "carga_horaria_minima": 1000,
    "estado_escolhido": "PI",
    "regiao_escolhida": "Nordeste",
    "instituicao_escolhida": "Colégio Técnico de Teresina",
    "regiao_metropolitana_ue": "NÃO"
}

# Exemplo de predição individual
def exemplo_predicao_individual():
    print("=== PREDIÇÃO INDIVIDUAL ===")
    response = requests.post(f"{BASE_URL}/predizer-evasao", json=dados_estudante)
    
    if response.status_code == 200:
        resultado = response.json()
        print(f"Probabilidade de Evasão: {resultado['probabilidade_evasao']:.2%}")
        print(f"Probabilidade de Não Evasão: {resultado['probabilidade_nao_evasao']:.2%}")
        print(f"Categoria de Risco: {resultado['categoria_risco']}")
        print(f"Mensagem: {resultado['mensagem']}")
    else:
        print(f"Erro: {response.status_code} - {response.text}")
```

```python
# Exemplo de predição em lote
def exemplo_predicao_lote():
    print("\n=== PREDIÇÃO EM LOTE ===")
    
    dados_lote = {
        "estudantes": [
            dados_estudante,
            {
                "cor_raca": "Branca",
                "idade": 18,
                "sexo": "Masculino",
                "renda_familiar": "1<RFP<=1,5",
                "modalidade_de_ensino": "Educação Presencial",
                "tipo_de_oferta": "Integrado",
                "turno": "Matutino",
                "nome_de_curso": "Técnico em Administração",
                "eixo_tecnologico_escolhido": "Gestão e Negócios",
                "carga_horaria_minima": 1200,
                "estado_escolhido": "SP",
                "regiao_escolhida": "Sudeste",
                "instituicao_escolhida": "IFSP - Campus São Paulo",
                "regiao_metropolitana_ue": "SIM"
            }
        ]
    }
    
    response = requests.post(f"{BASE_URL}/predizer-evasao-lote", json=dados_lote)
    
    if response.status_code == 200:
        resultados = response.json()
        for predicao in resultados['predicoes']:
            print(f"\nEstudante {predicao['indice'] + 1}:")
            print(f"  Curso: {predicao['estudante']['nome_curso']}")
            print(f"  Probabilidade de Evasão: {predicao['probabilidade_evasao']:.2%}")
            print(f"  Categoria: {predicao['categoria_risco']}")
    else:
        print(f"Erro: {response.status_code} - {response.text}")
```

```python
# Verificar saúde da API
def verificar_saude():
    print("=== VERIFICAÇÃO DE SAÚDE ===")
    response = requests.get(f"{BASE_URL}/health")
    print(f"Status: {response.json()}")

```

```python
if __name__ == "__main__":
    verificar_saude()
    exemplo_predicao_individual()
    exemplo_predicao_lote()
```

```python
# Célula 1: Definir o app
from fastapi import FastAPI

app = FastAPI()

@app.get("/modelo/info")
async def info():
    return {"message": "API funcionando!"}

# Célula 2: Executar com nest_asyncio (permite rodar em Jupyter)
import nest_asyncio
import uvicorn

nest_asyncio.apply()

# Executar em thread separada
def run_server():
    uvicorn.run(app, host="0.0.0.0", port=8000, log_level="info")

# Iniciar em background
import threading
thread = threading.Thread(target=run_server, daemon=True)
thread.start()

print("Servidor rodando em http://localhost:8000")
```
